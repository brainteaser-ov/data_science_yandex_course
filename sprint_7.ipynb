Компания предоставила данные с характеристиками сотрудников компании. Среди них — <mark style="background-color: #00FFFF">уровень удовлетворённости</mark> сотрудника работой в компании. Эту информацию получили из форм обратной связи: сотрудники заполняют тест-опросник, и по его результатам рассчитывается доля их удовлетворённости от 0 до 1, где <mark style="background-color: #00FFFF"><u>0 — совершенно неудовлетворён</u></mark>, <mark style="background-color: #00FFFF"><u>1 — полностью удовлетворён</u></mark>. <br> 
Собирать данные такими опросниками не так легко: компания большая, и всех сотрудников надо сначала оповестить об опросе, а затем проследить, что все его прошли. <br>

Почему бизнесу это важно: удовлетворённость работой напрямую влияет на отток сотрудников. А предсказание оттока — одна из важнейших задач HR-аналитиков. Внезапные увольнения несут в себе риски для компании, особенно если уходит важный сотрудник.<br>

```python
def analyze_dataframe(df):
    # Вывод первых 5 строк датафрейма
    print("\nHead:")
    display(df.head())

    # Вывод общей информации о датафрейме
    print("\nInfo:")
    display(df.info())

    # Вывод основных статистических характеристик датафрейма
    print("\nDescribe:")
    display(df.describe())

    # Приведение колонок к нижнему регистру
    df.columns = df.columns.str.lower()
    
    # Проверка наличия дубликатов в датафрейме
    if df.duplicated().sum() > 0:
        print("\nНайдены явные дубликаты!")
        df.drop_duplicates()
        print("\nЯвные дубликаты успешно удалены.")
    else:
        print("\nЯвных дубликатов не найдено.")
        
    # Проверка наличия пропусков в датафрейме
    if df.isna().sum().any():
        print("\nНайдены пропуски!")
    else:
        print("\nПропуски не найдены.")
    return df

display(analyze_dataframe(train))
display(analyze_dataframe(test))
display(analyze_dataframe(target))
```

```python
def count_empty_strings(df, column):
    empty_values = df[column].isin(['']).sum()
    return empty_values

display(count_empty_strings(test, 'dept'))
display(count_empty_strings(train, 'workload'))
```

```python
# Заменим пустую строку на nan
def replace_empty_strings(df, column):
    df[column] = df[column].replace(' ', float('nan'))
    return df
display(replace_empty_strings(test, 'dept'))
display(replace_empty_strings(test, 'workload'))
```

```python
# Проверим пропуски и nan
def calculate_missing_values(data):
  
    # Подсчет пропусков
    missing_values = pd.DataFrame(round((data.isna().mean()*100),2).sort_values(ascending=False), columns=['NaNs, %'])
    
    # Вывод общего количества пропусков
    total_missing = data.isna().sum().sum()
    total_cells = len(data) * len(data.columns)
    empty_percentage = total_missing / total_cells
    print('Процент пропусков: {:.0%}'.format(empty_percentage))
    
    # Проверка на наличие пропусков в столбцах
    columns_with_missing_values = data.columns[data.isna().any()].tolist()
    if columns_with_missing_values:
        print('Столбцы с пропусками:')
        for column in columns_with_missing_values:
            missing_percentage = round((data[column].isna().sum() / len(data)) * 100, 2)
            print(f'{column}: {missing_percentage}%')
    else:
        print('Все столбцы заполнены')
        
display(calculate_missing_values(train))
display(calculate_missing_values(test))
```

```python
def plot_num_correlation(X, y):
    
    X_numeric = X.select_dtypes(include=['int', 'float'])
   
    
    correlation_kendall = X_numeric.corrwith(y, method='kendall')
    correlation_spearman = X_numeric.corrwith(y, method='spearman')
    
    plt.figure(figsize=(12, 8))
    sns.heatmap(pd.concat([correlation_kendall, correlation_spearman], axis=1),
                annot=True, cmap='seismic', fmt=".2f",
                xticklabels=['Kendall', 'Spearman'],
                yticklabels=X_numeric.columns, cbar=False)

    correlated_columns = []
    for column in X_numeric.columns:
        corr_value = correlation_spearman[column]
        # Проверка на мультиколлинеарность
        if abs(corr_value) > 0.9 and abs(corr_value) < 0.95:
            print("Столбцы", column, "и", correlated_columns, "имеют мультиколлинеарность.")
        if abs(corr_value) > 0.3:
            correlated_columns.append(column)
        
        if abs(corr_value) < 0.3:
            category = "Нет корр."
        elif abs(corr_value) < 0.4:
            category = "Слабая"
        elif abs(corr_value) < 0.7:
            category = "Средняя"
        else:
            category = "Сильная"
        
        if corr_value < 0:
            category += " -"
        else:
            category += " +"
        
        plt.text(2.1, X_numeric.columns.get_loc(column) + 0.5, category, ha='center', va='center')
    
    plt.show()
    
    return correlated_columns



correlated_columns = plot_num_correlation(X_train, y_train)
display(correlated_columns)
```

```python
def calc_smape(test, pred):
    return 1/ len (test) * np.sum (2 * np.abs (pred-test) / (np.abs (test) + np.abs (pred))*100)
```

```python
def employment_years(train_quit):
    train_quit['employment_years'] = train_quit['employment_years'].astype(int)
    employment_years = train_quit[train_quit['quit'] == 'yes']['employment_years'].value_counts()
    
    # Словарь для хранения данных о продолжительности работы
    years_dict = {'Малая продолжительность': 0, 'Средняя продолжительность': 0, 'Долгая продолжительность': 0}
    
    # Процентное содержание каждой группы
    total_employees = employment_years.sum()
    
    for years, count in employment_years.items():
        if years <= 2:
            years_dict['Малая продолжительность'] += count
        elif years > 2 and years <= 6:
            years_dict['Средняя продолжительность'] += count
        elif years > 6 and years <= 10:
            years_dict['Долгая продолжительность'] += count
    for years, count in years_dict.items():
        percentage = (count / total_employees) * 100
        print(f'{years}: {percentage:.2f}%')

employment_years(train_quit)
```

&#x1F31F; Исходя из этих данных, можно разделить столбец "employment_years" на следующие категории:<br>
1. Менее 2 лет: сотрудники, которые работали в компании менее 2 лет.<br>
2. От 2 до 6 лет: сотрудники, которые работали в компании от 2 до 6 лет.<br>
3. Более 6 лет: сотрудники, которые работали в компании более 6 лет.<br>
