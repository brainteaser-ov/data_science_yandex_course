```python
def analyze_dataframe(data):
    
    missed_cells = data.isnull().sum().sum()/(data.shape[0]*(data.shape[1]-1))
    missed_rows = sum(data.isnull().sum(axis = 1)>0)/data.shape[0]
    print ('\033[1m' + '\nПроверка пропусков'+ '\033[0m')
    print ('Количество пропусков: {:.0f}'.format(data.isnull().sum().sum()))
    print ('Доля пропусков: {:.1%}'.format(missed_cells)+ '\033[0m')
    print ('Доля строк содержащих пропуски: {:.1%}'.format(missed_rows))
    
    ## Проверим дубликаты
    print ('\033[1m' + '\nПроверка на дубликаты'+ '\033[0m')
    print('Количество явных дубликатов: ', data.duplicated().sum())
    
    ## Посмотрим на сами данные
    print ('\033[1m' + '\nПервые пять строк датасета'+ '\033[0m')
    display(data.head())
    
    
    # Вывод основных статистических характеристик датафрейма
    
    print('\033[1m' + '\n0писание количественных данных:'+ '\033[0m')
    display(data.describe().T)
    
    print ('\033[1m' + '\nОписание категориальных данных:'+ '\033[0m')
    
    display(data.describe(include='object').T)
    print('\033[1m' + '\nВывод уникальных значений по каждому категориальному признаку:'+ '\033[0m')
    
    df_object = data.select_dtypes(include='object').columns
    
    for i in df_object:
        print('\033[1m' + '_'+ str(i) + '\033[0m')
        display(data[i].value_counts())
        
    # Вывод общей информации о датафрейме
    print('\033[1m' + '\nInfo:'+ '\033[0m')
    display(data.info())

    # Приведение колонок к нижнему регистру
    data.columns = data.columns.str.lower()
    
        
display(analyze_dataframe(data_0))
```

```python
def plots_numerics(df):
    df = df.drop('id', axis=1).copy()
  #  df = df.drop(columns=columns_list, axis=1).copy()
    column = df.select_dtypes(include=['int', 'float']).columns

    # Создание списка цветов для диаграмм
    colors = ['#00FFFF', '#FF00FF', '#ff355e', '#fc74fd', '#ff1493', '#66ff00', '#00cc99', '#CC00FF']
    color_cycle = itertools.cycle(colors)

    for col in column:
        color = next(color_cycle)
        n = len(df[col])
        bins = int(round(np.log2(n) + 1))

        # Визуализация распределения
        if not df[col].isnull().all():
            sns.histplot(data=df, bins=bins, x=col, color=color)
            plt.title(f"Распределение {col}", fontweight='bold')
            plt.show()

        # Визуализация выбросов
        sns.boxplot(data=df, x=col, color=color)
        plt.title(f"График {col}", fontweight='bold')
        plt.show()

    # Вывод подробной статистики
    display(df[column].describe())
```

```python
# Проверим границы выбросов
def calculate_outliers(df, column_name):
    q1 = df[column_name].quantile(0.25)
    q3 = df[column_name].quantile(0.75)
    iqr = q3 - q1
    lower_bound = round(q1 - 1.5 * iqr, 2)
    upper_bound = round(q3 + 1.5 * iqr, 2)

    outliers = df[(df[column_name] < lower_bound) | (df[column_name] > upper_bound)][column_name]
    
    #return lower_bound, upper_bound
    print("Нижняя граница выбросов:", lower_bound)
    print("Верхняя граница выбросов:", upper_bound)
```

```python
def plot_num_correlation(df, col):
    
    y = df[col]
    df = df.drop(col, axis=1)
    X_numeric = df.select_dtypes(include=['int', 'float'])
   
    correlation_kendall = X_numeric.corrwith(y, method='kendall')
    correlation_spearman = X_numeric.corrwith(y, method='spearman')
    
    plt.figure(figsize=(6, 6))
    sns.heatmap(pd.concat([correlation_kendall, correlation_spearman], axis=1),
                annot=True, cmap='seismic', fmt=".2f",
                xticklabels=['Kendall', 'Spearman'],
                yticklabels=X_numeric.columns, cbar=False)
    plt.title(f"Корреляция признаков с '{col}'", fontweight='bold')

    correlated_columns = []
    for column in X_numeric.columns:
        corr_value = correlation_spearman[column]
        # Проверка на мультиколлинеарность
        if abs(corr_value) > 0.9 and abs(corr_value) < 0.95:
            print("Столбцы", column, "и", correlated_columns, "имеют мультиколлинеарность.")
        if abs(corr_value) > 0.3:
            correlated_columns.append(column)
        
        if abs(corr_value) < 0.3:
            category = "Нет корр."
        elif abs(corr_value) < 0.4:
            category = "Слабая"
        elif abs(corr_value) < 0.7:
            category = "Средняя"
        else:
            category = "Сильная"
        
        if corr_value < 0:
            category += " -"
        else:
            category += " +"
        
        plt.text(2.2, X_numeric.columns.get_loc(column) + 0.5, category, ha='center', va='center')
    
    plt.show()
    
    sns.heatmap(df.corr(), vmin=-1, vmax=1, annot=True, cmap='viridis', cbar=False)
    plt.title(f"Корреляция признаков", fontweight='bold')

    plt.show()
    
    return correlated_columns



correlated_columns = plot_num_correlation(data_0, 'product')
display(correlated_columns)
```

## Обучение и проверка модели

```python
# напишем функции, необходимые для обучения модели
target_col = 'product'

# Обучение модели
def model_fit(df, name):
    df = df.drop('id', axis=1).copy()
    X = df.drop(target_col, axis=1)
    y = df[target_col]

    X_train, X_test, y_train, y_test = train_test_split(
        X, 
        y, 
        test_size=0.25, random_state=12345)

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    
    X_train = pd.DataFrame(X_train_scaled,columns=X.columns)
    X_test = pd.DataFrame(X_test_scaled, columns=X.columns)


    # обучение модели и получение предсказаний на тестовых данных
    model_lr  = LinearRegression()
    model_lr.fit(X_train, y_train)
    y_pred = model_lr.predict(X_test)
    mean_value = y_pred.mean()
    print('\033[1m' + f'*** {name} ***'+'\033[0m')
    print('Cредний запас предсказанного обьема: \t {:.2f} тыс. бареллей'.format(y_pred.mean()))
   
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)

    print(f'Среднеквадратичная ошибка (RMSE) модели: {rmse:.2f}')
   
    y_test = y_test.reset_index(drop=True) 
    predictions = pd.Series(y_pred)
    
    # добавим столбец с предсказаниями в датафреймы
    
    df['predict'] = model_lr.predict(df[['f0', 'f1', 'f2']])
    display(df.head())
    print('\n————————————————————————————————————————————————————————————')
    
    return predictions, y_test, mean_value, 
```

```python
y_pred_0, y_test_0, mean_value_0 = model_fit(data_0, 'Регион 1')
y_pred_1, y_test_1, mean_value_1 = model_fit(data_1, 'Регион 2')
y_pred_2, y_test_2, mean_value_2 = model_fit(data_2, 'Регион 3')
```

## Подготовка к расчёту прибыли

```python
TOTAL = 500 # количество точек, исследуемых при разведке региона (500)
BEST = 200 #количество лучших точек, выбранных для разработки (200)
BUDGET = 10000  # бюджет на разработку скважин в регионе (10 млрд рублей)
IPB  = 0.45 # income_per_barrel доход с одного барреля сырья (450 тыс. рублей)
risk_threshold = 0.025 # пороговое значение вероятности убытков (2.5%)
```


<div class="alert alert-block alert-success">
<b>Успех:</b> Лайк за константы в верхнем регистре, не забываешь про PEP8
</div> 

```python
# рассчитаем общий бюджет на разработку скважин в регионе:
well_budget = BUDGET / BEST
   
# оценим средний доход с одной скважины:
average_income = well_budget * IPB
  

# рассчитаем пороговое значение объема продукции, при котором вероятность убытков составит 2.5%:
threshold_volume = well_budget / IPB
```

```python
print(f'Достаточный объём сырья для безубыточной разработки новой скважины: {threshold_volume:.2f} тыс. баррелей')
```

```python
# сравним полученный объём сырья со средним запасом в каждом регионе
print("Первый регион:", mean_value_0 - threshold_volume)
print("Второй регион:", mean_value_1 - threshold_volume)
print("Третий регион:", mean_value_2 - threshold_volume)
```

```python
# функция расчёта прибыли по выбранным скважинам

def revenue(target, predictions):
    target = pd.Series(target)
    predictions = pd.Series(predictions, index=target.index)
    top_predictions = predictions.sort_values(ascending=False)
    selected = target[top_predictions.index][:BEST]

    return selected.sum() * IPB - BUDGET 
```

<b>Успех:</b> Ф-ция для подсчета прибыли определена верно
</div>

## Расчёт прибыли и рисков 

```python
state = np.random.RandomState(12345)
```

```python
# функция подсчёта средней выручки, доверительного интервала 95% и риска убытков
def bootstrap_count(target, predictions):
    values = []
    for i in range(1000):
        target = target.reset_index(drop=True)
       
        target_subsample = target.sample(TOTAL, replace=True, random_state=state)
        probs_subsample = predictions[target_subsample.index]
       # res = revenue(target_subsample, probs_subsample)
        rev = pd.Series(revenue(target_subsample, probs_subsample)).quantile(0.95)
        values.append(rev)
    
        

    values = pd.Series(values)
    lower = values.quantile(0.025)
    upper = values.quantile(0.975)
    loss = (values < 0).mean()
    

    mean = values.mean()
    
    print("Средняя выручка:", mean)
    print(f"95%-й доверительный интервал: {lower} | {upper}" )
    print(f'Риск убытков: {loss:.1%}\n')
```

<b>Успех:</b> С бутстрапом тоже все в порядке (сэмплирование происходит с возвращением), метрики считаются верным образом!  
</div>

<div class="alert alert-block alert-info">
    
<b>Совет: </b> Но `.quantile(0.95)` избыточен. Нам же нужно получить чистое распределение прибыли, и уже после мы используем квантили для подсчета доверительного интервала
