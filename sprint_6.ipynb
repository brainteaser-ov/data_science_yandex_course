```python
def change_data_types(df):
    # заменим запятые на точки 
    df = df.apply(lambda x: x.str.replace(',', '.') if x.dtype == 'object' else x)
    df['эке (энергетическая кормовая единица)'] = df['эке (энергетическая кормовая единица)'].astype(float)
    df['спо (сахаро-протеиновое соотношение)'] = df['спо (сахаро-протеиновое соотношение)'].astype(float)
    df['жирность,%'] = df['жирность,%'].astype(float)
    df['белок,%'] = df['белок,%'].astype(float)
 
    
    return df

df = change_data_types(df)
display(df.head())
```

```python
# Приведем все значения строк к нижнему регистру
def lowercase_columns(df):
    for column in df.columns:
        if df[column].dtype == 'object':
            df[column] = df[column].apply(lambda x: x.lower() if isinstance(x, str) else x)
    return df


lowercase_columns(df)
lowercase_columns(data)
lowercase_columns(cow_buy)
display(df.head())
display(data.head())
display(cow_buy.head())
```

```python
def rename_columns(df):
    df.rename(columns={'вкус молока': 'вкус_молока',
                       'тип пастбища': 'тип_пастбища',
                       'удой, кг': 'удой', 
                       'сырой протеин, г': 'сырой_протеин', 
                       'жирность,%': 'жирность', 
                       'белок,%': 'белок', 
                       'эке (энергетическая кормовая единица)': 'кормовая_единица', 
                       'спо (сахаро-протеиновое соотношение)': 'спо'}, inplace=True)
    df['порода папы_быка'] = df['порода папы_быка'].replace('айдиалл', 'айдиал')
    df['порода'] = df['порода'].replace('вис бик айдиалл', 'вис бик айдиал')
    df['тип_пастбища'] = df['тип_пастбища'].replace('равнинные', 'равнинное')


rename_columns(df)
display(df.head())
```

<span style="color:#8b51f6">Создание функций, необходимых для обучения модели, рассчета метрик и визуализации результатов</span> &#x1F4CC;

```python
# Напишем функции, необходимые для обучения модели, рассчета метрик и визуализации результутов

# Обучение модели
def model_fit(df):
    X = df.drop(target_col, axis=1)
    y = df[target_col]

    X_train, X_test, y_train, y_test = train_test_split(
        X, 
        y, 
        random_state=RANDOM_STATE)
    
    # подготовка признаков (масштабирование и кодирование)
    encoder = OneHotEncoder(drop='first', sparse=False)
    X_train_ohe = encoder.fit_transform(X_train[cat_col_names])
    X_test_ohe = encoder.transform(X_test[cat_col_names])

    encoder_col_names = encoder.get_feature_names_out(cat_col_names)

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train[num_col_names])
    X_test_scaled = scaler.transform(X_test[num_col_names])

    X_train_ohe = pd.DataFrame(X_train_ohe, columns=encoder_col_names)
    X_test_ohe = pd.DataFrame(X_test_ohe, columns=encoder_col_names)

    X_train_scaled = pd.DataFrame(X_train_scaled, columns=num_col_names)
    X_test_scaled = pd.DataFrame(X_test_scaled, columns=num_col_names)

    X_train = pd.concat([X_train_ohe, X_train_scaled], axis=1)
    X_test = pd.concat([X_test_ohe, X_test_scaled], axis=1)

    # обучение модели и получение предсказаний на тестовых данных
    model_lr  = LinearRegression()
    model_lr = model_lr.fit(X_train, y_train)
    predictions = model_lr.predict(X_test)
    
    return X, X_train, X_test, y_train, y_test, predictions, model_lr

# Рассчет метрик
def metrics_calculations(y_test, predictions):
   
    # средняя абсолютная ошибка на тестовых данных 
    mae = mean_absolute_error(y_test, predictions)

    mse = mean_squared_error(y_test, predictions)
    print(f'MSE = {mse:.0f}')

    rmse = np.sqrt(mse)
    print(f'RMSE = {rmse:.0f}')

    print(f'MAE = {mae:.0f}')
    
    # рассчет значения R2 для линейной регрессии
    r2_lr = r2_score(y_test, predictions)   
    print("R2 LinearRegression =", r2_lr) 

    residials = y_test - predictions
    residuals_var = np.var(residials)
    print(residuals_var)
    return residials
    
    
# Визуализация результатов 
def fabulous_plots(residials, X_train, predictions, y_test):

    # графики по количественным признакам тренировочного датасета:
    # гистограмма частотности распределения и ящик с усами.
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10,5))

    X_train[num_col_names].plot(kind='hist', bins=10, ax=axes[0])
    axes[0].set_title('Гистограмма распределения')

    X_train[num_col_names].plot(kind='box', ax=axes[1], rot=45)
    axes[1].set_title('Разброс значений признаков')

    plt.show()
    # # построение графиков
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10,5))
    axes[0].hist(residials)
    axes[0].set_title('Гистограмма распределения остатков')
    axes[0].set_xlabel('Остатки')

    axes[1].scatter(predictions, residials)
    axes[1].set_xlabel('Предсказания модели')
    axes[1].set_ylabel('Остатки')
    axes[1].set_title('Анализ дисперсии')
    plt.show()


    # Создаем список с номерами точек данных
    x = range(len(y_test))

    # Строим график для y_test
    plt.plot(x, y_test, label='y_test')

    # Строим график для predictions
    plt.plot(x, predictions, label='predictions')

    # Добавляем легенду
    plt.legend()

    # Добавляем подписи осей на русском языке
    plt.xlabel('Удой')
    plt.ylabel('Значение')

    # Отображаем график
    plt.show()
    print (X_train.shape)
```

```python
def calculate_precision(true_labels, predicted_labels):
    TN = []
    FN = []
    FP = []
    TP = []

    for true, pred in zip(true_labels, predicted_labels):
        if true == 0 and pred == 1:
            FP.append(pred)
        elif true == 1 and pred == 1:
            TP.append(pred)
        elif true == 0 and pred == 0:
            TN.append(pred)
        elif true == 1 and pred == 0:
            FN.append(pred)

    TN_count = len(TN)
    FN_count = len(FN)
    FP_count = len(FP)
    TP_count = len(TP)


    precision = TP_count / (TP_count + FP_count)

    return precision


print(calculate_precision(data_pr["y_valid"], data_pr["y_pred_0.1"]))
#print(calculate_precision(data_pr["y_valid"], ddata_pr["y_pred_0.11"]))
# print(calculate_precision(data_pr["y_valid"], data_pr["y_pred_0.125"]))
# print(calculate_precision(data_pr["y_valid"], data_pr["y_pred_0.133"]))
```
