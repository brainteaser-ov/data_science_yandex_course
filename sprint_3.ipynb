Исходя из расчетов, средняя продолжительность поездок составляет около 17.81 минуты, а медианная продолжительность - около 17.68 минуты. Минимальное значение продолжительности составляет 0.5 минуты, а максимальное - 40.82 минуты.<br>

На основе этих данных можно сделать следующие выводы:<br>

1. Средняя и медианная продолжительность поездок практически одинаковы, что может указывать на отсутствие значительных выбросов или аномалий в данных, однако на графике boxplot можно видеть наличие выбросов (точки, находящиеся за пределами усов). Выбросы выше верхнего максимума и ниже нижнего минимума могут указывать на наличие аномалий или ошибок в данных.
<br> 

2. Минимальное значение продолжительности 0.5 минуты может быть необычным, так как обычно ожидается, что поездки будут иметь более длительную продолжительность.<br>

3. Максимальное значение продолжительности 40.82 минуты также может быть необычным, так как это достаточно длительное время для поездки.<br>

В целом, данные показывают некоторые аномалии в столбце 'duration', такие как очень короткие и очень длинные поездки. Необходимо дополнительно исследовать эти аномалии и проверить их достоверность.<br>

```python
#Посчитаем количество значений от 0.5 до 1 и если их число не превышает 10% датасета заменим их медианой
def replace_duration(df):
    # Подсчет количества значений от 0.5 до 1
    count = df[(df['duration'] >= 0.5) & (df['duration'] <= 1)]['duration'].count()
    
    # Проверка условия на количество значений
    if count <= 0.1 * len(df):
        # Замена значений на медиану
        median_duration = df['duration'].median()
        df.loc[(df['duration'] >= 0.5) & (df['duration'] <= 1), 'duration'] = median_duration
    
    return df
rides_df = replace_duration(rides_df)
rides_df
```

```python
mean_duration = rides_df['duration'].mean()
median_duration = rides_df['duration'].median()
min_duration = rides_df['duration'].min()
max_duration = rides_df['duration'].max()
print("Mean duration:", mean_duration)
print("Median duration:", median_duration)
print("Minimum duration:", min_duration)
print("Maximum duration:", max_duration)
```

```python
def categorize_age(data):
    # Создаем новый столбец 'age_category' и заполняем его значениями по условиям
    age_category = []
    for age in data['age']:
        if age <= 18:
            age_category.append('дети и подростки')
        elif age <= 35:
            age_category.append('молодежь')
        else:
            age_category.append('средний возраст')
    
    data['age_category'] = age_category
    
    # Считаем количество значений в каждой категории
    category_counts = data['age_category'].value_counts()
    
    # Считаем процентное соотношение каждой категории от общего количества значений в столбце
    category_percentages = category_counts / len(data) * 100
    
    return category_percentages


result = categorize_age(users_df)
print(result.round())
```

```python
def categorize_distance(data):
    # Создаем новый столбец 'distance_category' и заполняем его значениями по условиям
    distance_category = []
    for distance in data['distance']:
        if distance <= 2000:
            distance_category.append('короткое расстояние')
        elif distance <= 4000:
            distance_category.append('среднее расстояние')
        else:
            distance_category.append('длинное расстояние')
    
    data['distance_category'] = distance_category
    
    # Считаем количество значений в каждой категории
    category_counts = data['distance_category'].value_counts()
    
    # Считаем процентное соотношение каждой категории от общего количества значений в столбце
    category_percentages = category_counts / len(data) * 100
    
    return category_percentages


result = categorize_distance(rides_df)
print(result.round())
```

```python
# Выборки продолжительности сессий для пользователей с подпиской и без подписки
duration_subscribers = users_with_subscription['duration']
duration_non_subscribers = users_without_subscription['duration']

# Вычисление средней продолжительности сессий для каждой группы
mean_duration_subscribers = duration_subscribers.mean()
mean_duration_non_subscribers = duration_non_subscribers.mean()

# Применение t-теста
t_statistic, p_value = stats.ttest_ind(duration_subscribers, duration_non_subscribers, alternative='greater')

# Вывод результатов
print("Средняя продолжительность сессий подписчиков:", mean_duration_subscribers)
print("Средняя продолжительность сессий не подписчиков:", mean_duration_non_subscribers)
print("t-статистика:", t_statistic)
print("p-значение:", p_value)
if p_value < 0.05:
    print("Разница в продолжительности сессий статистически значима")
else:
    print("Разница в продолжительности сессий не является статистически значимой")
```
