```python
def analyze_dataframe(df, column_names):
    # Вывод первых 5 строк датафрейма
    print("\nHead:")
    display(df.head())

    # Вывод общей информации о датафрейме
    print("\nInfo:")
    display(df.info())

    # Вывод основных статистических характеристик датафрейма
    print("\nDescribe:")
    display(df.describe())

    # Приведение колонок к нижнему регистру
    df.columns = df.columns.str.lower()
    
    # Приведение значений колонок к нижнему регистру с проверкой на NaN
    for column in column_names:
        if df[column].dtype == 'object':
            df[column] = df[column].apply(lambda x: x.lower() if pd.notnull(x) else x)
    else:
        print(f"\nВ колонке {column} изменение регистра не требуется!")

    
    # Проверка наличия дубликатов в датафрейме
    if df.duplicated(subset=column_names).sum() > 0:
        print("\nНайдены дубликаты!")
        df.drop_duplicates(subset=column_names, inplace=True)
        
        # Проверка оставшихся дубликатов после удаления
        if df.duplicated(subset=column_names).sum() > 0:
            print("\nОстались дубликаты после удаления!")
        else:
            print("\nДубликаты успешно удалены.")
    else:
        print("\nДубликаты не найдены.")
   
    return df

column_names = ['name', 'platform', 'year_of_release']
display(analyze_dataframe(data, column_names))
```

```python
# check
# пропущенные значения бары

def pass_value_barh(df):
    try:
        (
            (df.isna().mean()*100)
            .to_frame()
            .rename(columns = {0:'space'})
            .query('space > 0')
            .sort_values(by = 'space', ascending = True)
            .plot(kind = 'barh', figsize = (19,6), rot = -5, legend = False, fontsize = 16)
            .set_title('Пример' + "\n", fontsize = 22, color = 'SteelBlue')    
        );    
    except:
        print('пропусков не осталось :) или произошла ошибка в первой части функции ')
```

```python
def pass_value_barh1(df):
    try:
        missing_values = (df.isna().mean() * 100).to_frame().rename(columns={'space': 'Процент пропущенных значений'})
        missing_values = missing_values[missing_values['Процент пропущенных значений'] > 0].sort_values(by='Процент пропущенных значений', ascending=True)
        
        fig, ax = plt.subplots(figsize=(10, 6))
        ax.barh(missing_values.index, missing_values['Процент пропущенных значений'], color='steelblue')
        
        ax.set_xlabel('Процент пропущенных значений')
        ax.set_ylabel('Столбец')
        ax.set_title('Процент пропущенных значений в каждом столбце')
        
        for i, v in enumerate(missing_values['Процент пропущенных значений']):
            ax.text(v + 1, i, f'{v:.2f}%', color='black', fontsize=10)
        
        plt.show()
    except:
        print('Нет пропущенных значений 🙂 или произошла ошибка в первой части функции')
```

Виды рейтингов, присваиваемых Entertainment Software Rating Board (ESRB), включают:<br>

&#x1F31F; **"EC"** (Early Childhood) - Рекомендуется для детей младшего возраста (от 3 лет и старше). Игры с этим рейтингом предназначены для самых маленьких игроков и обычно содержат простые и ненасильственные материалы.<br>

&#x1F31F; Платформы с меньшими продажами (общие продажи менее 250 миллионов долларов): PC, GC, XOne, PS, WiiU, PSV, N64, GB, DC, WS<br>

```python
#Функция категоризации общего числа продаж с 2014 по 2016 гг.
def categorise_sales1(data):
    filtered_data = data.query('year_of_release >= 2014')
    platform_sales = filtered_data.groupby('platform')['total_sales'].sum().sort_values(ascending=False).reset_index()
    platform_sales.columns = ['name', 'total_sales']
    
    for index, row in platform_sales.iterrows():
        if row['total_sales'] > 100:
            platform_sales.at[index, 'category'] = 'крупные продажи'
        elif row['total_sales'] < 50:
            platform_sales.at[index, 'category'] = 'меньшие продажи'
        else:
            platform_sales.at[index, 'category'] = 'средние продажи'
    
    return platform_sales
display(categorise_sales1(data))
```

```python
# Визуализируем данные о продажах
def plot_sales(data):
    platform_sales = categorise_sales1(data)
    
    # Создаем фигуру и оси графика
    fig, ax = plt.subplots()
    
    # Задаем данные для графика
    x = platform_sales['name']
    y = platform_sales['total_sales']
    categories = platform_sales['category']
    
    # Задаем цвета для каждой категории
    colors = {'средние продажи': 'orange', 'меньшие продажи': 'red', 'крупные продажи': 'green'}
    
    # Создаем столбцы для каждой платформы
    bars = ax.bar(x, y)
    
    # Задаем цвета столбцов в соответствии с категориями
    for bar, category in zip(bars, categories):
        bar.set_color(colors[category])
    
    # Добавляем легенду
    legend_handles = [plt.Rectangle((0,0),1,1, color=color) for color in colors.values()]
    ax.legend(legend_handles, colors.keys())


    # Добавляем подписи осей и заголовок
    ax.set_xlabel('Платформа')
    ax.set_ylabel('Продажи')
    ax.set_title('Продажи по категориям')
    
    # Поворачиваем подписи оси x на 90 градусов, чтобы они не перекрывали друг друга
    plt.xticks(rotation=90)
    
    
    # Отображаем график
    plt.show()

# Вызываем функцию для создания и отображения графика
plot_sales(data)
```

```python
#функция для анализа срока продолжительности"жизни" платформ
def count_years(data):
    platform_years = data.groupby('platform')['year'].agg(['min', 'max'])
    platform_years['lifespan'] = platform_years['max'] - platform_years['min']
    # Создание столбца с категорией
    platform_years['category'] = ''
    for index, row in platform_years.iterrows():
        if row['lifespan'] <= 3:
            platform_years.loc[index, 'category'] = 'короткий'
        elif row['lifespan'] <= 13:
            platform_years.loc[index, 'category'] = 'средний'
        else:
            platform_years.loc[index, 'category'] = 'долгий'
    lifespan_table = platform_years.groupby('category').size().reset_index(name='count')
    
    life_sr = platform_years[platform_years['category'] == 'средний']
    for index, row in life_sr.iterrows():
        if row['lifespan'] <= 5:
            life_sr.loc[index, 'category'] = 'короткий'
        elif row['lifespan'] <= 9:
            life_sr.loc[index, 'category'] = 'средний'
        else:
            life_sr.loc[index, 'category'] = 'долгий'
    lifespan_tab = life_sr.groupby('category').size().reset_index(name='count')
    print("Таблица сгруппированных платформ по сроку 'жизни':")
    print(lifespan_table)
    print(platform_years)
    print(life_sr)
    print(lifespan_tab)
    average_lifespan = life_sr['lifespan'].mean()
    print("Средняя продолжительность жизни платформы:", round(average_lifespan, 2), "лет")

    plt.figure(figsize=(8, 8))
    life_sr['category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
    plt.title('Длительность "жизни" платформы')
    plt.ylabel('')
    plt.legend()
    plt.show()

display(count_years(data))
```

```python
def market_analysis(data):
    # Шаг 1: Фильтрация данных по году выпуска игры
    filtered_data = data.query('year_of_release >= 2013')
    
    # Шаг 2: Создание сводной таблицы с суммированием продаж по платформам и выбор пяти платформ с наибольшими продажами
    platform_sales = filtered_data.pivot_table(index='platform', values='total_sales', aggfunc='sum')
    top_5_platforms = platform_sales.nlargest(5, 'total_sales')
    
    # Шаг 3: Добавление категории "other" для всех остальных платформ
    other_sales = platform_sales.loc[~platform_sales.index.isin(top_5_platforms.index)].sum()
    top_5_platforms.loc['other'] = other_sales
    
    # Шаг 4: Построение круговой диаграммы для доли продаж каждой платформы
    fig, axs = plt.subplots(1, 3, figsize=(18, 6))
    
    # Шаг 5: Повторение шагов 1-4 для трех регионов: Северная Америка, Европа и Япония
    regions = ['na_sales', 'eu_sales', 'jp_sales']
    for i, region in enumerate(regions):
        region_data = filtered_data.pivot_table(index='platform', values=region, aggfunc='sum')
        top_5_platforms_region = region_data.nlargest(5, region)
        other_sales_region = region_data.loc[~region_data.index.isin(top_5_platforms_region.index)].sum()
        top_5_platforms_region.loc['other'] = other_sales_region
        
        plt.subplot(1, 3, i+1)
        plt.pie(top_5_platforms_region[region], labels=top_5_platforms_region.index, autopct='%1.1f%%')
        plt.title(f'Доля продаж платформ в {region.upper()}')
    
    plt.show()
```

#### Шаг 5. Проверка гипотез
<a id='cell7'></a>

<span style="color:#8b51f6">Средние пользовательские рейтинги платформ Xbox One и PC одинаковые</span> &#x1F4CC; 

**Нулевая гипотеза (H0)**: Средние пользовательские рейтинги платформ Xbox One и PC равны.<br>
**Альтернативная гипотеза (H1)**: Средние пользовательские рейтинги платформ Xbox One и PC различаются.<br>
Для проверки данной гипотезы можно использовать t-тест для независимых выборок. Этот критерий используется для сравнения средних значений двух независимых групп. В данном случае мы сравниваем средние пользовательские рейтинги двух разных платформ. T-тест является стандартным инструментом для проверки гипотез о различии средних значений в двух группах.<br>
Пороговое значение alpha можно задать, например, 0.05. Это означает, что если p-value (вероятность получить такие или еще более экстремальные результаты при условии верности нулевой гипотезы) будет меньше 0.05, то мы отвергнем нулевую гипотезу в пользу альтернативной.<br>

<div class="alert alert-success">
<font size="4", color= "seagreen"><b>✔️ Комментарий ревьюера</b></font>
    <br /> 
    <font size="3", color = "black">
<br /> Гипотезы сформулированы верно

```python
# Создаем две выборки пользовательских рейтингов для платформ Xbox One и PC
df = data.query('year_of_release >= 2013')
xbox_one_ratings = df[df['platform'] == 'xone']['user_score'].dropna()
pc_ratings = df[df['platform'] == 'pc']['user_score'].dropna()

# Вычисляем средние значения
mean_xbox_one = np.mean(xbox_one_ratings)
mean_pc = np.mean(pc_ratings)

# Выполняем t-тест для независимых выборок
alpha = 0.05 # Задаем пороговое значение alpha

results = stats.ttest_ind(xbox_one_ratings, pc_ratings)
p_value = results.pvalue

# Печатаем результаты
print("Средний пользовательский рейтинг Xbox One:", mean_xbox_one)
print("Средний пользовательский рейтинг PC:", mean_pc)
print("p-значение:", p_value) 

# Сравниваем полученное p-value с пороговым значением alpha
if p_value < alpha:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не отвергаем нулевую гипотезу")
```

После выравнивания периодов использования платформ на основании проведенного t-теста для независимых выборок, значение p-значения равно 0.14, что больше порогового значения alpha (0.05) и означает, что с вероятностью в 14% такое или большее различие можно получить случайно. <br>
Это означает, что мы не отвергаем нулевую гипотезу о равенстве средних пользовательских рейтингов платформ Xbox One и PC.<br>

<b>Из теории на тренажере</b>
        
Формулирование двусторонних гипотез. <br>
        
<b>Никакие экспериментально полученные данные никогда не подтвердят какую-либо гипотезу. Это наше фундаментальное ограничение. </b>Данные могут лишь не противоречить ей или, наоборот, показывать крайне маловероятные результаты (при условии, что гипотеза верна). Но и в том, и в другом случае нет оснований утверждать, что выдвинутая гипотеза доказана.
Допустим, данные гипотезе не противоречат, тогда мы её не отвергаем. Если же мы приходим к выводу, что получить такие данные в рамках этой гипотезы вряд ли возможно, у нас появляется основание отбросить эту гипотезу.
    
P-значение (англ. P-value) — величина, используемая при тестировании статистических гипотез. <b>Фактически это вероятность ошибки при отклонении нулевой гипотезы (ошибки первого рода)</b> пример ниже

```python
# Приведены два датасета: сумма покупок, совершённых за месяц посетителями ...

sample_1 = [3071, 3636, 3454, 3151, 2185, 3259, 1727, 2263, 2015,
2582, 4815, 633, 3186, 887, 2028, 3589, 2564, 1422, 1785,
3180, 1770, 2716, 2546, 1848, 4644, 3134, 475, 2686,
1838, 3352]
sample_2 = [1211, 1228, 2157, 3699, 600, 1898, 1688, 1420, 5048, 3007,
509, 3777, 5583, 3949, 121, 1674, 4300, 1338, 3066,
3562, 1010, 2311, 462, 863, 2021, 528, 1849, 255,
1740, 2596]
alpha = .05 # критический уровень статистической значимости
# если p-value окажется меньше него - отвергнем гипотезу
results = stats.ttest_ind(
sample_1,
sample_2)
print('p-значение:', results.pvalue)
if (results.pvalue < alpha):
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу")
```

<span style="color:#8b51f6">Средние пользовательские рейтинги жанров Action и Sports разные</span> &#x1F4CC; 

**Альтернативная гипотеза (H1)**: Средние пользовательские рейтинги жанров Action и Sports равны.<br>
**Нулевая гипотеза (H0)**: Средние пользовательские рейтинги жанров Action и Sports различаются.<br>
Для проверки данной гипотезы можно использовать t-тест для независимых выборок. Этот критерий используется для сравнения средних значений двух независимых групп. В данном случае мы сравниваем средние пользовательские рейтинги двух разных платформ. T-тест является стандартным инструментом для проверки гипотез о различии средних значений в двух группах.<br>
Пороговое значение alpha можно задать, например, 0.05. Это означает, что если p-value (вероятность получить такие или еще более экстремальные результаты при условии верности нулевой гипотезы) будет меньше 0.05, то мы отвергнем нулевую гипотезу в пользу альтернативной.<br>

```python
# Создание двух выборок с пользовательскими рейтингами жанров Action и Sports

action_ratings = df[df['genre'] == 'Action']['user_score'].dropna()
sports_ratings = df[df['genre'] == 'Sports']['user_score'].dropna()

# Проведение t-теста для независимых выборок
alpha = 0.05  # Задание порогового значения alpha

statistic, p_value = stats.ttest_ind(action_ratings, sports_ratings)

print("Статистика:", statistic)
print("p-значение:", p_value)

# Вывод результатов
if p_value < alpha:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не отвергаем нулевую гипотезу")
```

В данном случае, p-значение (1.0517832389140023e-27) иеньше уровня значимости, что означает, что вероятность получить такие различия случайно очень низкая (p-значение < 0.05), поэтому мы отвергаем нулевую гипотезу о равенстве средних пользовательских рейтингов.

<span style="color:#8b51f6">**Проверка гипотез**</span> &#x2705; <br>

<span style="color:#8b51f6">Средние пользовательские рейтинги платформ Xbox One и PC одинаковые</span> &#x1F4CC; <br>

**Нулевая гипотеза (H0)**: Средние пользовательские рейтинги платформ Xbox One и PC равны.<br>
**Альтернативная гипотеза (H1)**: Средние пользовательские рейтинги платформ Xbox One и PC различаются.<br>
Для проверки данной гипотезы можно использовать t-тест для независимых выборок. Этот критерий используется для сравнения средних значений двух независимых групп. В данном случае мы сравниваем средние пользовательские рейтинги двух разных платформ. T-тест является стандартным инструментом для проверки гипотез о различии средних значений в двух группах.<br>
Пороговое значение alpha можно задать, например, 0.05. Это означает, что если p-value (вероятность получить такие или еще более экстремальные результаты при условии верности нулевой гипотезы) будет меньше 0.05, то мы отвергнем нулевую гипотезу в пользу альтернативной.<br>
После выравнивания периодов использования платформ на основании проведенного t-теста для независимых выборок, значение p-значения равно 0.14012658403611647, что больше порогового значения alpha (0.05).<br>
Это означает, что мы не отвергаем нулевую гипотезу о равенстве средних пользовательских рейтингов платформ Xbox One и PC.<br>
Таким образом, у нас нет статистически значимых доказательств того, что средние пользовательские рейтинги платформ Xbox One и PC различаются.<br>

<span style="color:#8b51f6">Средние пользовательские рейтинги жанров Action и Sports разные</span> &#x1F4CC; <br>

**Альтернативная гипотеза (H1)**: Средние пользовательские рейтинги жанров Action и Sports равны.<br>
**Нулевая гипотеза (H0)**: Средние пользовательские рейтинги жанров Action и Sports различаются.<br>
Для проверки данной гипотезы можно использовать t-тест для независимых выборок. Этот критерий используется для сравнения средних значений двух независимых групп. В данном случае мы сравниваем средние пользовательские рейтинги двух разных платформ. T-тест является стандартным инструментом для проверки гипотез о различии средних значений в двух группах.<br>
Пороговое значение alpha можно задать, например, 0.05. Это означает, что если p-value (вероятность получить такие или еще более экстремальные результаты при условии верности нулевой гипотезы) будет меньше 0.05, то мы отвергнем нулевую гипотезу в пользу альтернативной.<br>
Исходя из результатов анализа независимых выборок средних пользовательских рейтингов жанров Action и Sports, мы получили  p-значение (1.0517832389140023e-27) меньше уровня значимости, что означает, что вероятность получить такие различия случайно очень низкая (p-значение < 0.05), поэтому мы отвергаем нулевую гипотезу о равенстве средних пользовательских рейтингов.

```python
data_games = pd.read_csv('/datasets/games.csv')
```

```python
data_games.columns = map(str.lower, data_games.columns)
```

```python
data_games = data_games.dropna(subset = ['year_of_release', 'name', 'genre'])
```

```python
data_games['user_score'] = data_games['user_score'].replace('tbd', np.nan).astype('float')
```

```python
data_games['rating'] = data_games['rating'].fillna('unknown')
```

```python
data_games['total_sales'] = data_games[['na_sales','eu_sales','jp_sales', 'other_sales']].sum(axis = 1)
```

```python
# check
# круги + категория другие

# функция не работает на версии библиотеки pandas = 2.0 (метод .append устарел)

def graph (df, year, region, name, axes):
    
    df = df.query('year_of_release >= @year')
    
    sales = df.pivot_table(index='platform', 
                           values=region, 
                           aggfunc='sum').nlargest(5, region)
    
    sales = sales.reset_index()
    
    sales = (
            sales.append({'platform': 'Other', region: df[region].sum() 
                       - sales[region].sum()}, ignore_index= True)
         )
    
    
    sales.columns = ['platform', 'sales']
      
# Цветовая гамма не подбиралась специально под платформы, кому-что попадется, 😉      
      
    labels_c=sales.platform
    colours = {'Wii':'C60', 'NES':'C1', 'GB':'C2', 'DS':'C3', 'X360':'C4', 
    'PS3':'C5', 'PS2':'C6', 'SNES':'C7', 'GBA':'C8',
               'PS4':'lightsteelblue', '3DS':'orange', 
               'N64':'C11', 'PS':'C12', 'XB':'C13', 'PC':'Fuchsia', '2600':'C15', 'PSP':'C48', 
               'XOne':'LimeGreen',
               'WiiU':'C18', 'GC':'C19', 'GEN':'C20', 'DC':'C21', 'PSV':'C22', 
               'SAT':'C23', 'SCD':'C24', 'WS':'C25', 'NG':'C26', 
               'TG16':'C27', '3DO':'C28', 'GG':'C29', 'PCFX':'C30', 'Other':'darkred'}
   
    sales.plot(kind='pie',
               y="sales",
               
               autopct='%1.0f%%',
               wedgeprops={'linewidth': 3.0, 'edgecolor': 'white'},
               textprops={'size': 'x-large'}, 
               labels= labels_c,
               colors=[colours[key] for key in labels_c],
               legend=False, 
               title = f"Популярность платформ в {name} ", 
               ax = axes).set(ylabel='')
    
    plt.tight_layout()
```

```python
# check
# круги в ряд

x_year = 2010

fig, axes = plt.subplots(1, 3, figsize = (15,6))
fig.suptitle(f'Обзор рынка платформ (портрет покупателя) с {x_year} г.  по 2016 г.', fontsize = 24, fontweight='bold')

graph(data_games, x_year, 'na_sales', 'Северной Америке', axes[0])
graph(data_games, x_year,'eu_sales', 'Европе', axes[1])
graph(data_games, x_year, 'jp_sales', 'Японии', axes[2])
```
